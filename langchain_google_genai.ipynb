```python
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI

api_key = userdata.get("GEMINI_API_KEY")

models = [
    "gemini-3.1-flash-lite",
    "gemini-3.5-flash-lite",
    "gemini-3.5-flash",
    "gemini-3.7-flash",
    "gemini-3.6-flash",
]

llms = [
    ChatGoogleGenerativeAI(model=m, google_api_key=api_key, temperature=0, max_retries=2)
    for m in models
]

llm = llms[0].with_fallbacks(llms[1:])
```

```python
response = llm.invoke(
    "Explain Python programming in simple English with an example."
)

print(response.content)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
```

```python
messages = [
    SystemMessage(
        content="You are a Python programming teacher. Explain everything in simple "
    ),
    HumanMessage(
        content="What is a Python list?"
    )
]
```

```python
response = llm.invoke(messages)

if isinstance(response.content, list):
    for block in response.content:
        if block.get("type") == "text":
            print(block["text"])
else:
    print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt = PromptTemplate.from_template(
    "Explain {topic} in simple English with an example."
)
```

```python
final_prompt = prompt.format(topic="Python")

print(final_prompt)
```

```python
final_prompt = prompt.format(topic="LangChain")

response = llm.invoke(final_prompt)

print(response.content)
```

```python
chain = prompt | llm
response = chain.invoke({
    "topic": "Retrieval Augmented Generation"
})

print(response.content)
```
