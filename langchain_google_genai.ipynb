```python
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI

api_key = userdata.get("GEMINI_API_KEY")

models = [
    "gemini-3.1-flash-lite",
    "gemini-3.5-flash-lite",
    "gemini-3.5-flash",
    "gemini-3.7-flash",
    "gemini-3.6-flash",
]

llms = [
    ChatGoogleGenerativeAI(model=m, google_api_key=api_key, temperature=0, max_retries=2)
    for m in models
]

llm = llms[0].with_fallbacks(llms[1:])
```

```python
response = llm.invoke(
    "Explain Python programming in simple English with an example."
)

print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an AI resume analyzer. Return only valid JSON. Do not include Markdown code fences."),
    ("human", """
    Extract candidate information from this text:

    {resume}

    Return these fields:
    - name
    - skills
    - experience_years
    """)
])
```

```python
from google.colab import userdata
from langchain_google_genai import GoogleGenerativeAIEmbeddings

api_key = userdata.get("GEMINI_API_KEY")

embeddings = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
    google_api_key=api_key
)

print("Embedding model initialized!")
```

```python
text = "I love learning Artificial Intelligence."

vector = embeddings.embed_query(text)

print(vector[:10])
print("Vector dimensions:", len(vector))
```

```python
texts = [
    "I love machine learning.",
    "I enjoy studying artificial intelligence.",
    "I like eating pizza."
]

vectors = embeddings.embed_documents(texts)

print("Number of vectors:", len(vectors))
print("Dimensions of first vector:", len(vectors[0]))
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
prompt= PromptTemplate.from_template( "Explain {topic} in simple English.")
```

```python
def count_words(text):
  word_count=len(text.split())
  return{"answer": text,
        "word_count": word_count
         }
```

```python
cleaning_runnable = RunnableLambda(count_words)
chain=(prompt | llm | StrOutputParser() | cleaning_runnable )
result = chain.invoke({
    "topic": "LangGraph"
})

print("Answer:", result["answer"])
print("Word count:", result["word_count"])
```

```python
from langchain_core.runnables import RunnableParallel
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
explanation_prompt=PromptTemplate.from_template("Explain {topic} in simple English.")
explanation_chain = explanation_prompt | llm | StrOutputParser()
```

```python
question_prompt=PromptTemplate.from_template("Generate one beginner-level interview question about {topic}")
question_chain = question_prompt | llm | StrOutputParser()
```

```python
keywords_prompt = PromptTemplate.from_template(
    "List five important keywords related to {topic}."
)

keywords_chain = keywords_prompt | llm | StrOutputParser()
```

```python
parallel_chain = RunnableParallel(
    explanation=explanation_chain,
    interview_question=question_chain,
    keywords=keywords_chain
)
```

```python
result = parallel_chain.invoke({
    "topic": "Python"
})
```

```python
print("Explanation:")
print(result["explanation"])

print("\nInterview Question:")
print(result["interview_question"])

print("\nKeywords:")
print(result["keywords"])
```
